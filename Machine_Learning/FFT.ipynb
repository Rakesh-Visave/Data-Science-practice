```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters for the signal
sampling_rate = 1000  # Sampling rate in Hz
T = 1.0 / sampling_rate  # Sampling interval in seconds
L = 1.0  # Length of the signal in seconds
N = int(L * sampling_rate)  # Number of samples

# Generate the time vector
t = np.linspace(0.0, L, N, endpoint=False)

# Generate a sample signal with two frequencies
freq1 = 50  # Frequency of the first component in Hz
freq2 = 150  # Frequency of the second component in Hz
signal = 0.5 * np.sin(2 * np.pi * freq1 * t) + 0.3 * np.sin(2 * np.pi * freq2 * t)

# Compute the FFT
yf = np.fft.fft(signal)
xf = np.fft.fftfreq(N, T)

# Plot the signal
plt.figure(figsize=(12, 6))

# Plot the time-domain signal
plt.subplot(2, 1, 1)
plt.plot(t, signal)
plt.title('Time Domain Signal')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')

# Plot the frequency-domain signal
plt.subplot(2, 1, 2)
plt.plot(xf[:N // 2], 2.0 / N * np.abs(yf[:N // 2]))
plt.title('Frequency Domain Signal')
plt.xlabel('Frequency [Hz]')
plt.ylabel('Amplitude')

plt.tight_layout()
plt.show()
```
