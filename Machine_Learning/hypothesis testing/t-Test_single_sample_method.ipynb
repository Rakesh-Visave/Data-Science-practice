```python
import numpy as np
import pandas as pd 
```

```python
train = pd.read_csv('train.csv').drop(columns=['Survived'])
test = pd.read_csv('test.csv')
```

```python
final_df = pd.concat([train,test]).sample(1309)
```

```python
pop = final_df['Age'].dropna()
```

```python
pop.mean()
```

```python
sample_age = pop.sample(25).values
```

```python
sample_age
```

```python
alpha = 0.05
```

```python
# H0 -> The mean age is 35
# H1 -> The mean is less than 35
```

```python
# check for normality using Shapiro Wilk test
from scipy.stats import shapiro
statistic,pvalue = shapiro(sample_age)
print(pvalue)
if pvalue > alpha:
    print('Data in Normal Distribution.')
else:
    print('Data in Not in Normal Distribution. ')
```

```python
pop_mean = 35
```

```python
import scipy.stats as stats

t_statistic, p_value = stats.ttest_1samp(sample_age, pop_mean)

print("t-statistic:", t_statistic)
print("p-value:", p_value/2)
```

```python


if p_value < alpha:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```
